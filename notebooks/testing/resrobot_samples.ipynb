```python
# Load probe output (robust path resolution)
import json
from pathlib import Path

candidates = [
    Path("resrobot-probe-output.json"),
    Path.cwd() / "resrobot-probe-output.json",
    Path.cwd() / "notebooks" / "testing" / "resrobot-probe-output.json",
    Path(__file__).resolve().parent / "resrobot-probe-output.json",
]
probe_path = next((p for p in candidates if p.exists()), None)

if not probe_path:
    print("Probe output not found. Checked:")
    for p in candidates:
        print(f"- {p}")
else:
    data = json.loads(probe_path.read_text(encoding="utf-8"))
    print(f"Probe run at: {data.get('runAt')}")
    results = data.get("results", [])
    for r in results:
        label = r.get("label")
        status = r.get("status")
        count = r.get("count")
        sample = r.get("sample") or []
        print(f"\n=== {label} ===")
        print(f"status={status}, count={count}")
        if sample:
            print("sample:")
            for s in sample:
                print(f"- {s}")
        else:
            print("sample: (none)")

```
